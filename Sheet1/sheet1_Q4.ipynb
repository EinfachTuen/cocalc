```python
def updateForVideo(a, function, x, line, ax):
    line.set_ydata(function(x, a))
    ax.set_title(fr"$a={a:.1f}$")
    return line,


def video_export(a_range, function, x, fig, ax, line):
    animation = FuncAnimation(
        fig,
        updateForVideo,
        frames=a_range,  
        fargs=(function, x, line, ax),
        interval=50
    )

    filename = "line_animation.mp4"
    animation.save(filename, writer="ffmpeg", fps=20)
    plt.close(fig)

    return Video(filename, embed=True)

# Accept either one number or an array of a-values
def plot_function(function, a, mode="plot"):
    x = np.linspace(-3, 3, 300)
    
    a_values = np.atleast_1d(a)

    fig, ax = plt.subplots(figsize=(6, 4))
    line, = ax.plot(x, function(x, a_values[0]))

    ax.set_xlim(-3, 3)
    ax.set_ylim(-10, 10)
    ax.set_xlabel("x")
    ax.set_ylabel("y")

    if mode == "video":
        return video_export(a_values, function, x, fig, ax, line)

    plt.show()

# test
def f(x, a):
    return x + a

plot_function(f, 3, "plot")

# function for excercises down
def f(x,a):
    return x**3 + a*x + 1
```

```python
from matplotlib.animation import FuncAnimation
from IPython.display import Video
import subprocess
import sys

try:
    import imageio_ffmpeg
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "--quiet", "imageio-ffmpeg"])
    import imageio_ffmpeg

plt.rcParams["animation.ffmpeg_path"] = imageio_ffmpeg.get_ffmpeg_exe()

x = np.linspace(-3, 3, 300)
a_values = np.linspace(-3, 3, 61)

fig, ax = plt.subplots(figsize=(6, 4))
line, = ax.plot(x, x + a_values[0])
ax.set_xlim(-3, 3)
ax.set_ylim(-10, 10)
ax.set_xlabel("x")
ax.set_ylabel(r"$x+a$")

def update(a):
    line.set_ydata(x + a)
    ax.set_title(fr"$a={a:.1f}$")
    return line,

animation = FuncAnimation(fig, update, frames=a_values, interval=50)
animation.save("/tmp/line_animation.mp4", writer="ffmpeg", fps=20)
plt.close(fig)
Video("/tmp/line_animation.mp4", embed=True)
```
