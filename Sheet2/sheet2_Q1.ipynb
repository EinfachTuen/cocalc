```python
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing()
median_income = housing.data[:, 0]
median_house_value = housing.target

print(housing.DESCR)

```

## (a)

Create a scatter plot of median house value against median income.

```python
# according to docu and info MedInc is in units of $10,000
# while target (house_value) is in $100,000.

plt.scatter(median_house_value, median_income)
plt.xlabel("median_house_value")
plt.ylabel("median_income")
plt.show()

median_income_usd      = median_income * 10_000
median_house_value_usd = median_house_value * 100_000

plt.scatter(median_house_value_usd,median_income_usd)
plt.ylabel("$_median_income")
plt.xlabel("$_median_house_value")    #
plt.show()
```
